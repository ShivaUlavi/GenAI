```python
import os
from langchain_openai import ChatOpenAI

# Verify Windows environment variables are detected
print("API Key Found:", "Yes" if os.getenv("AICREDITS_API_KEY") else "No")
print("Base URL:", os.getenv("AICREDITS_BASE_URL"))

# Quick test with GPT-4o
try:
    llm = ChatOpenAI(
        api_key=os.getenv("AICREDITS_API_KEY"),
        base_url=os.getenv("AICREDITS_BASE_URL"),
        model="anthropic/claude-fable-5"
    )
    print("Response:", llm.invoke("Say hi!").content)
    
except Exception as e:
    print("Error executing model:", e)
```
