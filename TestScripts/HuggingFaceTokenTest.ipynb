```python
import os
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from langchain_core.prompts import PromptTemplate

hf_token = os.environ.get("HUGGINGFACEHUB_API_TOKEN")

if not hf_token:
    print("❌ Error: HUGGINGFACEHUB_API_TOKEN is missing from your environment variables.")
else:
    print("✅ Token found! Initializing LangChain + Hugging Face Pipeline...")
    
    try:
        # 1. Base endpoint wrapped strictly as an inference container
        raw_llm = HuggingFaceEndpoint(
            repo_id="meta-llama/Llama-3.3-70B-Instruct",
            max_new_tokens=250,
            huggingfacehub_api_token=hf_token
        )
        
        # 2. Wrap it inside ChatHuggingFace to force conversational format automatically
        llm = ChatHuggingFace(llm=raw_llm)
        
        # 3. Standard LangChain prompt construction
        template = """You are a helpful AI assistant. 
Explain the core concept of {topic} using a simple analogy that a {audience} can understand."""

        prompt = PromptTemplate.from_template(template)
        
        # 4. Bind into an active LCEL Chain
        chain = prompt | llm
        
        print("\n🚀 Invoking the Hugging Face Conversational Chain...")
        print("-" * 60)
        
        # 5. Execute
        result = chain.invoke({"topic": "Vector Databases", "audience": "high school student"})
        
        print("🤗 LangChain + Hugging Face Output:\n")
        print(result.content.strip())
        
    except Exception as e:
        print(f"\n❌ Configuration Error: {e}")
```
