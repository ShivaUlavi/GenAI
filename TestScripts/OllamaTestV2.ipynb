```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate

print("🔄 Connecting to local Ollama instance...")

try:
    # 1. Initialize Gemma 2 using its exact tag from your 'ollama list'
    llm = ChatOllama(
        model="gemma2:2b",
        temperature=0.7
    )
    
    # 2. Build a structured LangChain Prompt Template
    template = """You are a helpful AI assistant running completely locally. 
Explain the core concept of {topic} using a simple analogy that a {audience} can understand."""

    prompt = PromptTemplate.from_template(template)
    
    # 3. Bind them together into an active LangChain Chain
    chain = prompt | llm
    
    print("\n🚀 Invoking your local Gemma 2 Chain...")
    print("-" * 60)
    
    # 4. Execute the pipeline (Runs completely offline off your D: Drive!)
    result = chain.invoke({"topic": "Quantum Computers", "audience": "high school student"})
    
    print("🤖 Local Gemma 2 Output:\n")
    print(result.content.strip())
    
except Exception as e:
    print(f"\n❌ Local Connection Error: {e}")
```
