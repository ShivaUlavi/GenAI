```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate

print("🔄 Connecting to local Ollama instance...")

try:
    # 1. Initialize the local model (make sure 'llama3' or your downloaded model name matches)
    llm = ChatOllama(
        model="llama3",
        temperature=0.7
    )
    
    # 2. Build the standard LangChain Prompt Template
    template = """You are a helpful local AI assistant. 
Explain the core concept of {topic} using a simple analogy that a {audience} can understand."""

    prompt = PromptTemplate.from_template(template)
    
    # 3. Bind into an active LangChain Chain
    chain = prompt | llm
    
    print("\n🚀 Invoking your local Ollama chain...")
    print("-" * 60)
    
    # 4. Execute completely locally without any API keys or internet!
    result = chain.invoke({"topic": "Quantum Computers", "audience": "high school student"})
    
    print("🦙 Local Ollama Output:\n")
    print(result.content.strip())
    
except Exception as e:
    print(f"\n❌ Local Connection Error: {e}")
    print("Make sure the Ollama application is actively running in your system tray and you have pulled the model via 'ollama run llama3'.")
```
