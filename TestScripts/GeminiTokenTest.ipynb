```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI

# 1. Verify your key is active
api_key = os.environ.get("GEMINI_API_KEY")

if not api_key:
    print("❌ Error: GEMINI_API_KEY is missing from your environment variables.")
else:
    print("✅ Key found! Connecting natively via LangChain...")
    
    try:
        # 2. Use the official modern class name that perfectly maps to the book's architecture
        llm = ChatGoogleGenerativeAI(
            model="gemini-3.6-flash",
            temperature=2.0,
            google_api_key=api_key
        )

        # 3. Test the basic prompt setup from the book
        response = llm.invoke("What is LangChain in one short sentence?")
        
        print("\n🤖 Official LangChain + Gemini Response:")
        print(response.content)
        
    except Exception as e:
        print(f"\n❌ Configuration Error: {e}")
```

```python
from langchain_core.prompts import PromptTemplate

# 1. Define a reusable template from your book structure
template = """You are a helpful AI assistant. 
Explain the core concept of {topic} using a simple analogy that a {audience} can understand."""

prompt = PromptTemplate.from_template(template)

# 2. Link the prompt and your working model together using a LangChain Chain
chain = prompt | llm

# 3. Dynamically inject your variables and invoke the chain
result = chain.invoke({"topic": "Quantum Computing", "audience": "high school student"})

```

```python
# Extract the text cleanly whether it returns a list or a string
if isinstance(result.content, list):
    # If it's a list of text blocks, grab the main text block
    clean_text = result.content[0].get('text', '') if isinstance(result.content[0], dict) else str(result.content[0])
else:
    clean_text = result.content

print("🤖 LangChain Unified Chain Output:\n")
print(clean_text.strip())
```
