```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
local_llm = ChatOllama(
    model="deepseek-r1:1.5b",
    temperature=0,)
```

```python
#creating an LCEL chain
prompt = PromptTemplate.from_template("Explain {concept} in simple terms")
local_chain = prompt | local_llm | StrOutputParser()
```

```python
result = local_chain.invoke({"concept": "quantum computing"})
print(result)
```

```python
result = local_chain.invoke({"concept": "God"})
print(result)
```

```python
from langchain_huggingface import ChatHuggingFace, HuggingFacePipeline
```

```python
from langchain_core.messages import SystemMessage, HumanMessage
```

```python
llmhf = HuggingFacePipeline.from_model_id(
    model_id="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    task="text-generation",
    pipeline_kwargs = dict(
        max_new_tokens=512,
        do_sample = False,
        repetition_penalty=1.03,
    ),
)
```

```python
chat_model = ChatHuggingFace(llm=llmhf)
```

```python
messages = [
    SystemMessage(content="You're a helpful assistant"),
    HumanMessage(
        content="Explain the concept of machine learning in simple terms"),]
ai_msg = chat_model.invoke(messages)
print(ai_msg.content)
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline
from langchain_huggingface import HuggingFacePipeline

model_id = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"

# 1. Load the tokenizer and model natively (No LangChain involved yet)
print("Loading model weights natively...")
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id)

# 2. Setup the text generation pipeline on the CPU
native_pipeline = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=512,
    do_sample=False,
    repetition_penalty=1.03
)

# 3. Pass the running pipeline directly into LangChain
llmhf = HuggingFacePipeline(pipeline=native_pipeline)
print("Pipeline wrapped in LangChain successfully!")
```
