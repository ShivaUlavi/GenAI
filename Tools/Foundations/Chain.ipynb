```python
from langchain_core.runnables import RunnableLambda
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
import os

api_key = os.getenv("AICREDITS_API_KEY")
base_url = os.getenv("AICREDITS_BASE_URL")
```

```python
llm = ChatOpenAI(
    model_name = "gpt-4o",
    api_key=api_key,
    base_url=base_url,
    temperature=0
)
```

```python
prompt = PromptTemplate.from_template("What is the capital of {input}?")
```

```python
chain = prompt | llm
```

```python
result = chain.invoke({"input": "Australia"})
```

```python
print(result.content)
```
