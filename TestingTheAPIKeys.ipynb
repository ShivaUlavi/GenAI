```python
import os
from langchain_openai import ChatOpenAI

# 1. Map your Gemini key to the endpoint configuration
api_key = os.environ.get("GEMINI_API_KEY")

if not api_key:
    print("❌ Error: GEMINI_API_KEY is missing from the environment.")
else:
    print("✅ Key found! Connecting via the Gemini-OpenAI bridge...")
    
    try:
        # 2. Point LangChain to Google's official compatibility gateway
        llm = ChatOpenAI(
            model="gemini-1.5-flash",
            openai_api_key=api_key,
            openai_api_base="https://googleapis.com"
        )

        # 3. Invoke the model as usual
        response = llm.invoke("What is LangChain in one short sentence?")
        
        print("\n🤖 Gemini Response:")
        print(response.content)
        
    except Exception as e:
        print(f"\n❌ Error: {e}")
```

```python
import os

try:
    # 1. Install the official Google GenAI SDK directly from the notebook
    print("Installing the official Google GenAI SDK...")
    %pip install -q --no-cache-dir google-genai
    
    from google import genai
    
    # 2. Check your active environment key
    api_key = os.environ.get("GEMINI_API_KEY")
    if not api_key:
        print("❌ Error: GEMINI_API_KEY is missing from the environment.")
    else:
        print("✅ Key found! Connecting directly to Google API...")
        
        # 3. Initialize the official client
        client = genai.Client() # It automatically picks up your GEMINI_API_KEY
        
        # 4. Request using the standard flash model
        response = client.models.generate_content(
            model='gemini-3.6-flash',
            contents='What is LangChain in one short sentence?',
        )
        
        print("\n🤖 Gemini Response:")
        print(response.text)

except Exception as e:
    print(f"\n❌ Error: {e}")
```
